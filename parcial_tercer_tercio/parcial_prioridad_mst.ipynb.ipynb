```python
import math
import matplotlib.pyplot as plt
# =========================================================
# PUNTO 1: COLA DE PRIORIDAD (MAX-HEAP)
# =========================================================
tareas = [
   ("T01", "Configurar servidor", 6),
   ("T02", "Reparar sensor", 10),
   ("T03", "Actualizar software", 4),
   ("T04", "Revisar cámara", 8),
   ("T05", "Generar respaldo", 3),
   ("T06", "Configurar alarma", 7),
]

def max_heapify(arr, n, i):
   mayor = i
   izq = 2 * i + 1
   der = 2 * i + 2
   if izq < n and arr[izq][2] > arr[mayor][2]:
       mayor = izq
   if der < n and arr[der][2] > arr[mayor][2]:
       mayor = der
   if mayor != i:
       arr[i], arr[mayor] = arr[mayor], arr[i]
       max_heapify(arr, n, mayor)

def build_max_heap(arr):
   h = list(arr)
   n = len(h)
   for i in range(n // 2 - 1, -1, -1):
       max_heapify(h, n, i)
   return h

def imprimir_arbol(h):
   if not h:
       print("Árbol vacío")
       return
   n = len(h)
   niveles = math.floor(math.log2(n)) + 1
   idx = 0
   for nivel in range(niveles):
       cant = min(2 ** nivel, n - idx)
       margen = " " * (2 ** (niveles - nivel + 1))
       sep = " " * (2 ** (niveles - nivel + 2) - 4)
       fila = margen
       for _ in range(cant):
           cod, _, prio = h[idx]
           fila += f"{cod}({prio}){sep}"
           idx += 1
       print(fila)

print("=== PUNTO 1 ===")
print("Arreglo inicial:")
for t in tareas:
   print(" ", t)
heap_tareas = build_max_heap(tareas)
print("\nMax-Heap resultante:")
for t in heap_tareas:
   print(" ", t)
print("\nÁrbol correspondiente al Heap:")
imprimir_arbol(heap_tareas)

# =========================================================
# PUNTO 2: PROCESAR TAREAS (EXTRACT-MAX E INSERCIÓN)
# =========================================================
def extract_max(h):
   if not h:
       return None
   tope = h[0]
   ultimo = h.pop()
   if h:
       h[0] = ultimo
       max_heapify(h, len(h), 0)
   return tope

def insertar_heap(h, elem):
   h.append(elem)
   pos = len(h) - 1
   while pos > 0:
       padre = (pos - 1) // 2
       if h[pos][2] > h[padre][2]:
           h[pos], h[padre] = h[padre], h[pos]
           pos = padre
       else:
           break

print("\n=== PUNTO 2 ===")
for i in range(2):
   procesada = extract_max(heap_tareas)
   print(f"\nTarea procesada #{i + 1}: {procesada}")
   print("Heap resultante:")
   for t in heap_tareas:
       print(" ", t)
insertar_heap(heap_tareas, ("T07", "Falla en sistema contra incendios", 12))
print("\nHeap después de insertar T07:")
for t in heap_tareas:
   print(" ", t)
print("\nÁrbol después de insertar T07:")
imprimir_arbol(heap_tareas)

# =========================================================
# PUNTO 3: DISPOSITIVOS A CONECTAR Y DISTANCIAS
# =========================================================
puntos = {
   "A": ("Servidor", 2, 2),
   "B": ("Cámara", 7, 3),
   "C": ("Sensor", 4, 7),
   "D": ("Alarma", 10, 8),
   "E": ("Router", 13, 3),
   "F": ("Panel", 15, 7),
}

def dist(u, v):
   _, x1, y1 = puntos[u]
   _, x2, y2 = puntos[v]
   return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("\n=== PUNTO 3 ===")
nodos = list(puntos.keys())
aristas = []
for i in range(len(nodos)):
   for j in range(i + 1, len(nodos)):
       u, v = nodos[i], nodos[j]
       w = dist(u, v)
       aristas.append((w, u, v))
       print(f"{u} - {v}: {w:.2f}")
plt.figure(figsize=(8, 5))
for k, (nom, x, y) in puntos.items():
   plt.scatter(x, y, s=120, zorder=3)
   plt.text(x, y + 0.35, f"{k}: {nom} ({x},{y})", ha="center", fontsize=8)
plt.title("Posición de los seis dispositivos")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

# =========================================================
# PUNTO 4: ÁRBOL DE EXPANSIÓN MÍNIMA (KRUSKAL MANUAL)
# =========================================================
padre = {n: n for n in nodos}
rango = {n: 0 for n in nodos}

def buscar(x):
   while padre[x] != x:
       padre[x] = padre[padre[x]]
       x = padre[x]
   return x

def unir(a, b):
   ra, rb = buscar(a), buscar(b)
   if ra == rb:
       return False
   if rango[ra] < rango[rb]:
       padre[ra] = rb
   elif rango[ra] > rango[rb]:
       padre[rb] = ra
   else:
       padre[rb] = ra
       rango[ra] += 1
   return True

aristas_ordenadas = sorted(aristas)
mst = []
for w, u, v in aristas_ordenadas:
   if unir(u, v):
       mst.append((u, v, w))
       if len(mst) == len(nodos) - 1:
           break
total_cable = sum(w for _, _, w in mst)
print("\n=== PUNTO 4 ===")
print("Conexiones seleccionadas:")
for u, v, w in mst:
   print(f"{u}{v}: {w:.2f}")
print(f"Longitud total de cable: {total_cable:.2f}")

# =========================================================
# PUNTO 5: VISUALIZACIÓN FINAL DEL MST
# =========================================================
plt.figure(figsize=(8, 5.5))
for u, v, w in mst:
   _, x1, y1 = puntos[u]
   _, x2, y2 = puntos[v]
   plt.plot([x1, x2], [y1, y2], "b-", lw=2)
   plt.text((x1 + x2) / 2, (y1 + y2) / 2 + 0.25, f"{w:.2f}", ha="center", fontsize=9)
for k, (nom, x, y) in puntos.items():
   plt.scatter(x, y, c="red", s=130, zorder=3)
   plt.text(x, y - 0.55, f"{k} ({nom})", ha="center", fontsize=8)
plt.title(f"MST Final - Longitud total de cable: {total_cable:.2f}")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
