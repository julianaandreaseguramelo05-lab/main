Una plataforma de streaming está analizando el crecimiento de nuevos suscriptores.
Se ha observado que el número de nuevos suscriptores en el mes n sigue el siguiente
comportamiento:
• En el mes 0 se registraron 0 nuevos usuarios.
• En el mes 1 se registró 1 nuevo usuario.
• A partir del mes 2, el número de nuevos usuarios es igual a la suma de los nuevos usuarios de
los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y
recomendaciones acumuladas del mes previo.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscriptores(n, memo={}):

   if n in memo:

       return memo[n]

   if n == 0:

       return 0

   if n == 1:

       return 1

   memo[n] = suscriptores(n - 1, memo) + suscriptores(n - 2, memo)

   return memo[n]

# Validación de entrada con bucle while

while True:

   entrada = input("Ingrese el número del mes solo números entero mayores a 0: ")

   if entrada.isdigit():

       n = int(entrada)

       break

   print("Error: Entrada inválida. Debe ingresar solo dígitos numéricos sin letras ni símbolos.\n")

# Cálculo del resultado

resultado = suscriptores(n)

print(f"En el mes {n} habrá {resultado} nuevos suscriptores.")
```
